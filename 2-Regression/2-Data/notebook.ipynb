```python
import pandas as pd
#Read the US pumpkin CSV file
pumpkins = pd.read_csv('../data/US-pumpkins.csv')

#filer out any rows that dont use price per bushel
pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel',case=True, regex=True)]
pumpkins.head()
```

```python
#Count the number of empty columns

pumpkins.isnull().sum()
```

```python
# Define the columns we need to keep
new_columns = ['Package', 'Month', 'Low Price', 'High Price', 'Date']

#Drop all the other columns
pumpkins = pumpkins.drop([c for c in pumpkins.columns if c not in new_columns], axis=1)
```

```python
# Avg prices
price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2

# Get month from the date column
month = pd.DatetimeIndex(pumpkins['Date']).month

# Create a new dataframe
new_pumpkins = pd.DataFrame({
    'Month': month, 'Package': pumpkins['Package'], 'Low Price': pumpkins['Low Price'], 'High Price': pumpkins['High Price'], 'Price': price})
```

```python
#convert the price of all cells prices by 1 1/9 bushels by dividing by 1/9

new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] = new_pumpkins.loc[new_pumpkins['Package'].str.contains('1 1/9'), 'Price'] / (1 + 1/9)

#convert the price of all cells prices by 1/2 bushels by dividing by 1/2
new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] = (
    new_pumpkins.loc[new_pumpkins['Package'].str.contains('1/2'), 'Price'] / 0.5
)
```

```python
import matplotlib.pyplot as plt

# getting inputs we need
price = pumpkins.price
month = pumpkins.month

# plotting the needed graph
plt.scatter(price,month)
plt.show()

```
